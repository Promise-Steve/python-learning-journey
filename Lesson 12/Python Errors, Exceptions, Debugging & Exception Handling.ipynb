## ---  Hospital Patient Data Validator

```python
# ---  handson exercise 1
def validate_patient_data():
    # --- this line validates patient name
    name = input("Enter Patient Name: ").strip()
    if not name:
        raise ValueError("Patient name cannot be empty. ")

    # --- for this line patient ID is validated
    patient_id = input("Enter patient_id: ").strip()
    if not patient_id:
        raise ValueError("Patient ID cannot be empty. ")
    
    # --- Age validation
    try:
        age_input = input("Enter Age: ")
        age = int(age_input)
        if  age < 0:
            raise ValueError("Age cannot be negative. ")
    except ValueError as e:
        if"cannot be negative" in str(e):
            raise e
        else:
            raise ValueError("Invalid age format. Please enter a whole number. ")

        # Validate weight
    try:
        weight_input = input("Enter weight (kg): ")
        weight = float(weight_input)
        if weight <= 0:
            raise ValueError("weight must be greater than zero.")
    except ValueError as e:
        if "greater than zero" in str(e):
            raise e
        else:
            raise ValueError("Invalid weight format. Please enter a number. ")

    # --- result
    print("\n========== PATIENT RECORD ==========")
    print(f"Patient Name: {name}")
    print(f"Patient ID: {patient_id}")
    print(f"Age: {age}")
    print(f"Weight: {weight} kg")
    print("Status: Validated")
    print("=======================================")
        
try:
    validate_patient_data()
except ValueError as error:
    print(f"\n[Validation Error]: {error}")
```

## --- E-Commerce Order Processing System

```python
# --- 

def main():
    print("========= E-COMMERCE ORDER SYSTEM ==========")

    # --- Requesting customer and product names
    customer_name = input("Enter Customer Name: ").strip()
    product_name = input("Enter Product Name: ").strip()

    # --- Validate that names are not empty
    if not customer_name or not product_name:
        print("\nError: Customer Name and Product Name cannot be empty.")
        return

    try:
        # --- product pricce request and conversion
        price_input = input("Enter Product Price (₦)): ")
        product_price = float(price_input)
        if product_price < 0:
            raise ValueError("Price cannot be negative. ")
        # --- quantity request and conversion
        quantity_input = input("Enter Quantity: ")
        quantity = int(quantity_input)

        # --- validate quantity boundaries
        if  quantity == 0:
            raise ValueError("Quantity must be greater than zero.")
        if quantity < 0:
            raise ValueError("Quantity cannot be negative.")
            
    except ValueError as e:
        # Catching conversion errors and custom raised errors
        print("\n========= ORDER ERROR ===========")
        if "could not convert" in str(e):
            print(f"Error: {e}")
        else:
            print(f"Error: {e}")
        print("Order Status: Rejected")
        print("===================================")
        return
    # --- calculate total cost
    subtotal = product_price * quantity
    discount = 0.0

    # --- Applying 10% discount if total >= ₦100,000
    if subtotal >= 100000:
        discount = subtotal * 0.10

    total_cost = subtotal - discount

    print("\n========= ORDER SUMMARY ==========")
    print(f"Customer: {customer_name}")
    print(f"Product: {product_name}")
    print(f"Unit Price: ₦{product_price:,.2f}")
    print(f"Quantity: {quantity}")

    if discount > 0:
        print(f"Subtotal: ₦{subtotal:,.2f}")
        print(f"Discount: ₦{discount:,.2f} (10% Applied)")

    print(f"Total Cos: ₦{total_cost:,.2f}")
    print("Order Status: Approved")
    print("==================================")

if __name__ == "__main__":
    main()
```

# --- Banking Transaction Validator

```python
def main():
    # --- Predefined account balance and transaction fee (Bonus Challenge)
    balance = 150000.00
    TRANSACTION_FEE = 100.00
    
    print("========== BANKING TRANSACTION SYSTEM ==========")
    print(f"Current Available Balance: ₦{balance:,.2f}")
    print(f"Standard Transaction Fee:   ₦{TRANSACTION_FEE:,.2f}")
    print("================================================")
    
    try:
        # --- Request withdrawal amount
        withdrawal_input = input("Enter withdrawal amount (₦): ")
        
        # --- Convert input to a numeric value
        withdrawal_amount = float(withdrawal_input)
        
        # --- Reject zero or negative withdrawals
        if withdrawal_amount == 0:
            raise ValueError("Withdrawal amount must be greater than zero.")
        if withdrawal_amount < 0:
            raise ValueError("Withdrawal amount cannot be negative.")
            
        # --- Calculate total required funds (Withdrawal + Fee)
        total_deduction = withdrawal_amount + TRANSACTION_FEE
        
        # --- Reject withdrawals greater than available balance 
        if total_deduction > balance:
            raise ValueError(
                f"Insufficient funds. Total deduction including fee is ₦{total_deduction:,.2f}, "
                f"which exceeds your balance of ₦{balance:,.2f}."
            )

    except ValueError as e:
        # --- Catching conversion errors and custom raised errors
        print("\n========== TRANSACTION FAILED ==========")
        if "could not convert" in str(e):
            print("Error: Invalid amount. Please enter a valid number.")
        else:
            print(f"Error: {e}")
        print("Transaction Status: Rejected")
        print("========================================")
        return

    # --- Calculate the remaining balance
    remaining_balance = balance - total_deduction

    # --- Display successful transaction summary
    print("\n========== TRANSACTION SUMMARY ==========")
    print(f"Previous Balance:   ₦{balance:,.2f}")
    print(f"Withdrawal Amount:  ₦{withdrawal_amount:,.2f}")
    print(f"Transaction Fee:    ₦{TRANSACTION_FEE:,.2f}")
    print(f"Total Deducted:     ₦{total_deduction:,.2f}")
    print("-----------------------------------------")
    print("Transaction Status: Successful")
    print(f"Remaining Balance:  ₦{remaining_balance:,.2f}")
    print("=========================================")

if __name__ == "__main__":
    main()
```
