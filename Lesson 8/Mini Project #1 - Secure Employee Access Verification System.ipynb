```python
# --- ==========================================
# --- Project: Study Buddy - SecureTech Solutions
# --- Program: Employee Access Verification System
# --- ==========================================

# --- Store Employee Information (Variables)
employee_name = "Grace Cletus"
age = 29
employment_status = "Active"
department = "Cybersecurity"
security_clearance = 4

print(f"Employee name: {employee_name}")
print(f"Age: {age}")
print(f"Employment Status: {employment_status}")
print(f"Department: {department}")
print(f"Security Clearance: {security_clearance}")
print("-" * 28)

# --- Verify Security Requirements
# --- Valid departments now include IT, Cybersecurity, or Executive
is_valid_department = (
    department == "IT" or department == "Cybersecurity" or department == "Executive"
)

if (
    age >= 18
    and employment_status == "Active"
    and is_valid_department
    and security_clearance >= 3
):
    
    print(f"Employee: {employee_name}\n")
    print("Access Granted\n")
    print("Security Level:")

# --- Determine and display the specific security level
    if security_clearance == 3:
        print("Standard Secure Access")
    elif security_clearance == 4:
        print("Advanced Secure Access")
    else:
        print("Top Secret Access")

else:
    print(f"Employee: {employee_name}\n")
    print("Access Denied")
```
