### --- HANDS-ON 1: User Message Collector

```python
# --- Print an introductory title for the tool
print("--- Interactive Message Collector ---")

while True:
    # --- Capture user input
    message = input("Enter a message or Q to quit: ")

    # --- Study Buddy Challenge: Convert input to uppercase before checking.
    # --- This catches 'q' and 'Q', and .strip() prevents accidental spaces from breaking the check.
    if message.strip().upper() == "Q":
        print("\nProgram terminated successfully. Goodbye!")
        break  # --- Immediately exits the loop, skipping the print statement below

    # --- Display the message only if it wasn't the quit command
    print(message)
    print()  # --- Prints an empty line for cleaner spacing in the terminal
```

#### --- HANDS-ON 2: Simple Command-Line Support System

```python
print("--- Command-Line Support System ---")

while True:
    # --- Capture user input and normalize it (lowercase & strip whitespace)
    command = input("Enter a command: ").strip().lower()

    # --- Decision system structure (if / elif / else)
    if command == "quit":
        print("Closing system...")
        break  # --- Terminate the loop immediately

    elif command == "help":
        # --- Study Buddy Challenge: Added 'name' and 'version' to the help list
        print("Available commands: help, status, name, version, quit")

    elif command == "status":
        print("System is running.")

    # --- Study Buddy Challenge: Command 1 (name)
    elif command == "name":
        print("System Name: PyAssist Support Desk")

    # --- Study Buddy Challenge: Command 2 (version)
    elif command == "version":
        print("Python Support System v1.0")

    else:
        # --- Fallback for any input that doesn't match the valid commands above
        print("Unknown command.")

    print()  # --- Adds a blank line for visual clarity between cycles
```

#### --- HANDS-ON 3: Simple Data Entry System

```python
print("--- Simple Sales Data Entry System ---")

# --- Initialize a variable to track total sales before the loop starts
total_sales = 0

while True:
    # --- Capture user input as text
    user_input = input("Enter sales amount or 0 to finish: ").strip()

    # --- Convert text input to an integer for numerical comparison
    amount = int(user_input)

    # --- Immediate exit check before processing the data
    if amount == 0:
        break  # --- Exits the loop completely, skipping the tracking and printing steps below

    # --- Update the total sales inside the loop for every valid entry
    total_sales += amount

    # --- Display the valid sales record
    print(f"Sales amount entered: {amount}")
    print()  # --- Adds a blank line for readability

# --- Final messages displayed after the loop has terminated
print("\nData entry completed.")
print(f"Total sales: {total_sales}")
```

#### --- The Controlled Loop Challenge

```python
print("--- Student Study Tracker ---")

while True:
    # --- Get user input and normalize it (lowercase & strip whitespace)
    subject = input("What did you study today? ").strip().lower()

    # --- Immediate exit check (Is it "quit"?)
    if subject == "quit":
        print("Study session completed.")
        break  # ---Terminate the loop instantly, exiting the program

    # --- Match recognized subjects
    elif subject == "python":
        print("Great! Keep practicing Python.")

    elif subject == "sql":
        print("Great! Keep practicing SQL.")

    elif subject == "excel":
        print("Great! Keep practicing Excel.")

    # --- Fallback for unrecognized subjects
    else:
        print("Subject not recognized. Try again.")

    print()  # --- Adds an empty line for clean visual spacing between questions
```
