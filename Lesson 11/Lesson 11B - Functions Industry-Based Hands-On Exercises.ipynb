#### --- Exercise 1: Employee Salary Processing System

```python
# --- Function to calculate Gross Salary
def calculate_gross_salary(basic_salary, housing, transport, other):
    """Calculates and returns the total gross salary."""
    return basic_salary + housing + transport + other


# --- Bonus Challenge: Function to calculate Net Salary
def calculate_net_salary(gross_salary, deductions):
    """Subtracts deductions from gross salary to return net salary."""
    return gross_salary - deductions


# --- Coach Timothy's Challenge: Handling multiple employees efficiently using a list of dictionaries
employees = [
    {
        "name": "Employee 1",
        "basic": 300000,
        "housing": 80000,
        "transport": 50000,
        "other": 20000,
        "deductions": 30000,
    },
    {
        "name": "Employee 2",
        "basic": 350000,
        "housing": 90000,
        "transport": 60000,
        "other": 20000,
        "deductions": 45000,
    },
    {
        "name": "Employee 3",
        "basic": 250000,
        "housing": 70000,
        "transport": 45000,
        "other": 25000,
        "deductions": 20000,
    },
]

# --- Process and display payroll for all employees dynamically
print("--- HR Payroll Processing Report ---")
for emp in employees:
    # --- Calculate Gross Salary
    gross = calculate_gross_salary(
        emp["basic"], emp["housing"], emp["transport"], emp["other"]
    )

    # --- Calculate Net Salary (Bonus)
    net = calculate_net_salary(gross, emp["deductions"])

    # --- Display the Results
    print(f"{emp['name']} | Gross Salary: {gross} | Net Salary: {net}")
```

### --- Exercise 2: Hospital Patient Record Formatter

```python
def format_patient_name(name):
    """Cleans and consistently formats patient names, handling empty inputs safely."""
    # --- Remove leading and trailing whitespace
    cleaned_name = name.strip()

    # --- Check if the name is empty after removing whitespace
    if not cleaned_name:
        return "Patient name cannot be empty."

    # --- Capitalize the first letter of each word (Title Case)
    return cleaned_name.title()


# --- List of test cases (including messy formats and empty inputs)
test_patients = [
    "   john doe   ",
    "MARY ANN JOHNSON",
    "  peter okafor",
    "",  # Empty string test
    "   ",  # String with only spaces test
]

# --- Process and display the patient records dynamically
print("--- Hospital Patient Record Cleaning Report ---")
for original_name in test_patients:
    formatted_name = format_patient_name(original_name)

    # --- Display both original and formatted names clearly
    print(f"Original:  '{original_name}'")
    print(f"Formatted: {formatted_name}")
    print("-" * 40)
```

```python
# --- Function to calculate discount based on percentage
def calculate_discount(subtotal, discount_percentage):
    """Calculates and returns the currency amount saved from a percentage discount."""
    return subtotal * (discount_percentage / 100)


# --- Function to calculate the final order total
def calculate_order_total(price, quantity, discount_percentage):
    """Calculates subtotal, determines discount amount, and returns final total."""
    subtotal = price * quantity

    # --- Leverage the discount calculation function
    discount_amount = calculate_discount(subtotal, discount_percentage)

    final_total = subtotal - discount_amount
    return final_total


# --- Scalable structure using a list of dictionaries
orders = [
    {"id": "Order 1", "price": 25000, "quantity": 3, "discount_pct": 6.6667},  # ~5000 discount
    {"id": "Order 2", "price": 15000, "quantity": 4, "discount_pct": 5},
    {"id": "Order 3", "price": 60000, "quantity": 2, "discount_pct": 0},
    {"id": "Order 4", "price": 8000, "quantity": 4, "discount_pct": 10},
    {"id": "Order 5", "price": 45000, "quantity": 2, "discount_pct": 4.722},
]

# --- Process and display e-commerce orders dynamically
print("--- E-Commerce Order Processing Report ---")
for order in orders:
    total_cost = calculate_order_total(
        order["price"], order["quantity"], order["discount_pct"]
    )

    # --- Display the result formatted to standard Naira currency format
    print(f"{order['id']} Total: ₦{total_cost:,.2f}")
```
