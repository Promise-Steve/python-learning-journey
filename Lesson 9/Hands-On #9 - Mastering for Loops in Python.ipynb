## --- HANDS - ON EXERCISE 2

# --- ATM Login Security System (Banking)

```python
successful = True

# --- Simulate up to 3 login attempts
for attempt in range(1, 4):
    print(f"Login Attempt {attempt}")
    
    if successful:
        print("Login Successful")
        break  # Exit the loop immediately
```

# --- BONUS CHALLENGE

```python
successful = False

# --- Simulate up to 3 login attempts
for attempt in range(1, 4):
    print(f"Login Attempt {attempt}")
    
    if successful:
        print("Login Successful")
        break
else:
    # --- This runs only if the loop completes all iterations without breaking
    print("\nAccount Locked.")
```

## --- HANDS-ON EXERCISE 3

# --- WAREHOUSE INVENTORY SCANNER (RETAIL)

```python
products = [
    "Laptop",
    "Keyboard",
    "Mouse",
    "Printer",
    "Monitor"
]

# --- Loop directly through each item in the product list
for product in products:
    print(f"Scanning {product}")

# --- Message displayed after the loop finishes
print("\nInventory scan completed successfully.")
```

# --- BONUS CHALLENGE

```python
products = [
    "Laptop",
    "Keyboard",
    "Mouse",
    "Printer",
    "Monitor"
]

# --- Use enumerate to get both the count and the product name
for index, product in enumerate(products, start=1):
    print(f"Product {index}: {product}")

# --- Message displayed after the loop finishes
print("\nInventory scan completed successfully.")
```

```python
# --- Outer loop controls the rows (1 to 5)
for row in range(1, 6):
    # --- Inner loop controls the seats in each row (1 to 6)
    for seat in range(1, 7):
        print(f"Row {row} - Seat {seat}")
```

# --- EMPLOYEE PAYROLL VERIFICATION (HUMAN RESOURCES)

```python
# --- Initialize a counter variable
processed_count = 0

# --- Loop through employee IDs from 1 to 20, skipping by 2 to get even numbers
for employee_id in range(2, 21, 2):
    print(f"Processing Employee ID {employee_id}")
    processed_count += 1  # --- Add 1 to the counter

# --- Print the final summary
print(f"\nTotal Employees Processed: {processed_count}")
```

```python
# --- Display the welcome message
print("=====================================")
print(" LifeCare Hospital Monitoring System ")
print("=====================================")
print()  # --- Adds an empty line for spacing

# --- Initialize a counter for blood pressure checks
bp_checks_count = 0

# --- Use a for loop to monitor 10 patients
for patient in range(1, 11):
    print(f"Checking Patient {patient}")
    
    # --- Check if the patient number is even for the BP check
    if patient % 2 == 0:
        print(f"Patient {patient} - Blood Pressure Checked")
        bp_checks_count += 1  # --- Increment the counter
        print()  # --- Adds spacing after the BP check to match expected output

# --- Display the total count summary
print(f"Total Blood Pressure Checks: {bp_checks_count}")
print()

# --- Final completion message
print("Daily Patient Monitoring Completed.")
```

```python
print("=====================================")
print(" LifeCare Hospital Monitoring System ")
print("=====================================")
print()

# --- Trackers for our final summary
total_patients_checked = 0
bp_checks_count = 0
critical_patients_detected = 0

# --- List of vital signs to check for each patient using a nested loop
vitals = ["Temperature", "Pulse", "Blood Pressure"]

# --- Loop through 10 patients
for patient in range(1, 11):
    print(f"Checking Patient {patient}")
    total_patients_checked += 1
    
    # --- Check if the patient number is divisible by 5 (Critical Patient)
    if patient % 5 == 0:
        print(f" ALERT: Patient {patient} is a CRITICAL PATIENT!")
        critical_patients_detected += 1
        
        # --- Stop monitoring early if a Critical Patient is encountered
        print("Stopping monitor early to assist critical patient...")
        print()
        break
        
    # --- Nested loop to check all 3 vital signs for normal patients
    for vital in vitals:
        # --- Normal flow: Only check Blood Pressure if the patient number is even
        if vital == "Blood Pressure" and patient % 2 != 0:
            continue  # --- Skip BP for odd-numbered, non-critical patients
            
        print(f"  -> Checking {vital}... Done.")
        
        # --- Track BP checks if it was actually performed
        if vital == "Blood Pressure":
            print(f"  -> Patient {patient} - Blood Pressure Checked")
            bp_checks_count += 1
            
    print()  # --- Visual spacing between patients

# --- Final Summary Report
print("=====================================")
print("           SUMMARY REPORT            ")
print("=====================================")
print(f"Total Patients Checked:        {total_patients_checked}")
print(f"Total Blood Pressure Checks:   {bp_checks_count}")
print(f"Critical Patients Detected:    {critical_patients_detected}")
print("=====================================")
print("Daily Patient Monitoring Completed.")
```
