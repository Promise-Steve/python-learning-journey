### --- Hands-On Exercise 1

## --- Banking & FinTech — Transaction Processing Counter

```python
# --- creating a variable named transaction

transaction = 1
successful_transactions = 0

while transaction <= 10:
    print(f"Processing Transaction {transaction}")

    successful_transactions += 1 # --- update successful transaction counts
    transaction += 1 # --- update the counter

# --- Bonus Challenge (display transaction counts)
print(f"\nStatus: All 10 transactions have been processed successfully")

# --- Extra challenge(display successfule transaction counts)
print(f"Successful_transaction: {successful_transactions}") 
```

### --- Hands-On Exercise 2

## --- Customer Service — Customer Input Validation System

```python
# --- variable initialization
attempts = 0
max_attempts = 3
valid_age = False

while not valid_age and attempts < max_attempts:
    age = int(input("Enter your age: "))
    attempts += 1 # This updates the attempt counter each time the user enters an age
    if 18 <= age <= 60:
        valid_age = True
    else:
        if attempts < max_attempts:
            print("Invalid age. Please enter an age between 18 and 60. \n")
if valid_age:
    print("Age accepted, Registration can continue.")
    print(f"Number of attempts: {attempts}")
else:
    print("Maximum attempts reached. Registration cannot continue.")
```

### --- Hands-On Exercise 3

## --- Retail & E-Commerce — Inventory Monitoring System

```python
# --- create variable

inventory = 50
orders_processed = 0

print("Initial inventory:", inventory, "units\n")

while inventory > 0:
    order = int(input("Enter order quantity: "))

# --- Check if order is larger than remaining inventory    
    if order > inventory:
        print("Insufficient inventory. ")
        print("Available:", inventory, "units")
        print("Requested:", order, "units\n")
    else:
# --- Process the Order (Subtract from inventory)
        inventory -= order
# --- Increase the processed orders counter
        orders_processed += 1
        
# --- Display the Remaining Inventory
        print("Order processed:", order, "units")
        print("Remaining inventory:", inventory, "units")
# --- Low-stock warning
        if inventory < 10 and inventory > 0:
            print("WARNING: Inventory is running low!")

        print() # ---blank line for clean spacing

# --- Final Summary when inventory hits 0
print("Inventory procesing completed. ")
print("Orders processed:", orders_processed)
print("Remaining inventory:", inventory, "units")
```
