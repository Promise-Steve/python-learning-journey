### --- Hands-On Exercise 1

## --- E-Commerce Product Combination Generator

```python
# Starting data
sizes = ["Small", "Medium", "Large"]
colors = ["Black", "White", "Blue"]

# Additional Challenge: Create a counter for total variations
total_variations = 0

# Create the index variable for the outer loop (sizes)
i = 0

# Outer while loop to move through sizes
while i < len(sizes):
    
    # Create/Reset the index variable for the inner loop (colors)
    j = 0
    
    # Inner while loop to move through colors
    while j < len(colors):
        # Print each size-color combination
        print(sizes[i], "-", colors[j])
        
        # Additional Challenge: Count this variation
        total_variations = total_variations + 1
        
        # Update the inner loop counter
        j = j + 1
        
    # Print a blank line for clean spacing between size categories
    print()
    
    # Update the outer loop counter
    i = i + 1

# Additional Challenge: Display the total number of product variations
print("Total Product Variations:", total_variations)
```

# --- Banking Transaction Risk Checker

```python
# Starting data
customers = ["C001", "C002", "C003"]
transactions = [45000, 125000, 75000]
risk_threshold = 100000

# Additional Challenge: Initialize counters
review_count = 0
total_comparisons = 0

# --- Create the index variable for the outer loop (customers)
i = 0

# --- Outer while loop to move through each customer
while i < len(customers):
    
    # --- Create/Reset the index variable for the inner loop (transactions)
    j = 0
    
    # --- Inner while loop to check all transactions against the current customer
    while j < len(transactions):
        
        # --- Additional Challenge: Track every comparison made
        total_comparisons = total_comparisons + 1
        
        # --- Business Rule: Check if the transaction exceeds the threshold
        if transactions[j] > risk_threshold:
            print("Customer:", customers[i], "| Transaction:", transactions[j], "| Status: REVIEW")
            
            # --- Additional Challenge: Track the flagged transaction
            review_count = review_count + 1
            
        # --- Update the inner loop counter
        j = j + 1
        
    # --- Update the outer loop counter
    i = i + 1

# --- Display final challenge metrics after loops finish
print("\nTransactions Requiring Review:", review_count)
print("Total Comparisons Performed:", total_comparisons)
```

# --- Hospital Appointment Scheduling Matrix

```python
# --- Starting data
doctors = ["Dr. Ahmed", "Dr. Grace", "Dr. Michael"]
time_slots = ["9:00 AM", "11:00 AM", "2:00 PM", "4:00 PM"]

# --- Additional & Expert Challenge: Initialize counters
total_appointments = 0
appointment_id = 1

# --- Create the index variable for the outer loop (doctors)
i = 0

# --- Outer while loop to move through each doctor
while i < len(doctors):
    
    # --- Create/Reset the index variable for the inner loop (time slots)
    j = 0
    
    # --- Inner while loop to move through all time slots for the current doctor
    while j < len(time_slots):
        
        # --- Expert Challenge: Print the combination with its unique ID
        print("Appointment", appointment_id, ":", doctors[i], "-", time_slots[j])
        
        # --- Additional Challenge: Count this possible appointment slot
        total_appointments = total_appointments + 1
        
        # --- Expert Challenge: Increase the ID number for the next slot
        appointment_id = appointment_id + 1
        
        # --- Update the inner loop counter
        j = j + 1
        
    # --- Print a blank line for clean spacing between different doctors
    print()
    
    # --- Update the outer loop counter
    i = i + 1

# --- Additional Challenge: Display the final summary metrics
print("Total Possible Appointments:", total_appointments)
```
