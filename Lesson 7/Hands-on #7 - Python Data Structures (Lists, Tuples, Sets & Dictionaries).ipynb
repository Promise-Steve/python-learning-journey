## --- Hospital Patient Registration System

### --- HANDS-ON EXERCISE 5

## --- BUSINESS SALES PERFORMANCE DASHBOARD

```python
# --- # Store the sales figures inside a list
sales = [2500, 3200, 4100, 2800, 3900, 3500]

# --- Display the initial list
print("Initial Sales List:", sales)
print()

# --- Display performance metrics using built-in functions
print("Total number of branches:", len(sales))
print("Highest sales:", max(sales))
print("Lowest sales:", min(sales))
print("Total sales:", sum(sales))
print()

# --- Add a new branch with sales of 4700
sales.append(4700)

# --- Display the updated list and new total branch count
print("Updated Sales List:", sales)
print("New total number of branches:", len(sales))
```
