```python
# This line of codes demonstrates managing inventorymanagement, logistics and warehouse systems
stock = int(input("initial_stock: "))
sold = int(input("no_sold: "))
arrived = int(input("no_arrived: "))
carton_capacity = int(input("qty_per_carton: "))

print("\n--- inventory report ---")

# 2. Update and display stock after sales
stock -= sold
print(f"stock_balance: {stock}")

# 3. Update and display stock after restocking
stock += arrived
stock += arrived
print(f"stock balance after restocking: {stock}")

# 4. Calculate full cartons using floor division
cartons = stock
cartons //= carton_capacity
print(f"number of full cartons: {cartons}")

# 5. Calculate remaining loose items using modulo assignment
loose_items = stock
loose_items %= carton_capacity
print(f"Remaining loose items: {loose_items}")

```
