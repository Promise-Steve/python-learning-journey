```python
# --- Exercise 2 focuses on using the f-string to generate automatically generate a apyroll receipt for a HR department
# --- Varaible creation
employee_name = "David Musa"
department = "Finance"
basic_salary = 185000
bonus = 35000
tax_rate = 0.075 # --- 7.5% expressed as decimal

# --- aritmethic calculation
gross_salary = basic_salary + bonus
tax_amount = gross_salary * tax_rate
net_salary = gross_salary - tax_amount
```

```python
# ---Output display
payroll_summary = f"""================================
        PAYROLL SUMMARY
================================


Employee     : {employee_name}

Departmen    : {department}

Basic Salary : ₦{basic_salary:.2f}

Bonus        : ₦{bonus:.2f}

Tax Rate     : {tax_rate}%

Gross Salary : ₦{gross_salary:.2f}

Tax Amount   : ₦{tax_amount:.2f}

Net Salary   : ₦{net_salary:.2f}


================================"""
print(payroll_summary)
```

```python
# ---  Exercise 3 demonstrates generating an online store receipt using the f-string
# --- Create variables
customer_name = "Amina Bello"
product = "Wireless Headphones"
quantity = 3
unit_price = 27500.00

# --- arithmetic calculations
total_cost = quantity * unit_price


# --- (quantity > 2) evaluates to True (1) if true, or False (0) if false
discount_multiplier = int(quantity > 2) 
discount_amount = total_cost * 0.10 * discount_multiplier
final_amount = total_cost - discount_amount

# --- Display the professionally formatted output
receipt = f"""====================================

      ONLINE STORE RECEIPT

====================================


Customer     : {customer_name}

Product      : {product}

Quantity     : {quantity}

Unit Price   : ₦{unit_price:,.2f}

Total Cost   : ₦{total_cost:,.2f}

Discount     : ₦{discount_amount:,.2f}

Final Amount : ₦{final_amount:,.2f}

Thank you for shopping with us!


===================================="""

print(receipt)
```

```python
# --- Create variables
employee_id = "SBC-2026-084"
employee_name = "Promise Steve"
department = "Data Analytics"
position = "Data Engineer"
years_of_experience = 4
monthly_salary = 450000.00
performance_rating = "Excellent"  # Options: Excellent, Very Good, Good, Needs Improvement

# --- arithmetic calculations
annual_base_salary = monthly_salary * 12
bonus_amount = annual_base_salary * 0.05
total_annual_package = annual_base_salary + bonus_amount

# --- Dynamic Status Mapping
status_lookup = {
    "Excellent": "Eligible for Promotion",
    "Very Good": "Outstanding Contribution",
    "Good": "Keep Improving",
    "Needs Improvement": "Performance Review Required"
}
hr_action_message = status_lookup.get(performance_rating, "Status Unknown")

# --- Generate the structured corporate profile summary
employee_profile = f"""==================================================

          SMARTBIZCRUX DATA SOLUTIONS
          
==================================================
[ EMPLOYEE IDENTITY PROFILE ]
--------------------------------------------------
Employee ID  : {employee_id}
Full Name    : {employee_name}
Department   : {department}
Current Role : {position}

[ PERFORMANCE & EXPERIENCE ]
--------------------------------------------------
Experience   : {years_of_experience} Years
HR Rating    : {performance_rating}
HR Action    : {hr_action_message}

[ COMPENSATIONS & PAYROLL ]
--------------------------------------------------
Monthly Base : ₦{monthly_salary:,.2f}
Annual Base  : ₦{annual_base_salary:,.2f}
Annual Bonus : ₦{bonus_amount:,.2f} (5%)
Total Gross  : ₦{total_annual_package:,.2f}
=================================================="""

print(employee_profile)
```
