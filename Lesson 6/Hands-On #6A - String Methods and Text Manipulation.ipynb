```python
# --- extending the handson oractice with the .format() function
first_name = ("Grace")
last_name = ("Johnson")
age = ("34 Years")
blood_group = ("O+")
department = ("Cardiology")

print("============ CAREPLUS HOSPITAL ============\n")
print("Patient Name:{}\n".format(first_name, last_name))
print("Age         :{}\n".format(age))
print("Blood Group :{}\n".format(blood_group))
print("Department  :{}\n".format(department))
print("=============================================")
```

```python
# --- extending the handson practicals using the f-string
# --- Creating variables
first_name = ("Grace")
last_name = ("Johnson")
age = ("34 Years")
blood_group = ("O+")
department = ("Cardiology")

print("============ CAREPLUS HOSPITAL ============\n")
print(f"Patient Name: {first_name.upper()} {last_name.upper()}\n")
print(f"Age         : {age}\n")
print(f"Blood Group : {blood_group}\n")
print(f"Department  : {department}\n")
print("=============================================")
```
