EMPLOYEE RECORD INFORMATION EXTRACTOR (HUMAN RESOURCES)

```python
# --- Creating variables

employee_record = "EMP-HR-2026-004512"

# --- Extracting specific parts

prefix = employee_record[0:3]
department = employee_record[4:6]
year = employee_record[7:11]
number = employee_record[12:18]

# --- slicing operations
first_10 = employee_record[:10]
last_6 = employee_record[-6:]
every_second = employee_record[::2]
every_third = employee_record[::3]
reversed_record = employee_record[::-1]

print(f"Employee Prefix       : {prefix}")
print(f"Department            : {department}")
print(f"Employment Year       : {year}")
print(f"Employee Number       : {number}")
print(f"First 10 characters   : {first_10}")
print(f"Last 6 Characters     : {last_6}")
print(f"Every Second Character: {every_second}")
print(f"Every Third Character : {every_third}")
print(f"Reverse               : {reversed_record}")
```

BANKING CARD SECURITY FORMATTER

```python
# --- Creating variable

card_number = "5489123476549876"

# --- Basic slicing
first_four = card_number[:4]
last_four = card_number[-4:]
middle_eight = card_number[4:12]

# --- Step and reverse slicing
every_second = card_number[::2]
every_third = card_number[::3]
reversed_card = card_number[::-1]
last_eight_reversed = card_number[:-9:-1]  # Steps backward from the last digit through the 8th

# --- Data Masking
masked_card = "**** **** **** " + card_number[-4:]


print(f"First 4 digits                    : {first_four}")
print(f"Last 4 digits                     : {last_four}")
print(f"Middle 8 digits                   : {middle_eight}")
print(f"Every second digit                : {every_second}")
print(f"Every third digit                 : {every_third}")
print(f"Entire card number in reverse     : {reversed_card}")
print(f"The last 8 digits in reverse order: {last_eight_reversed}")
print(f"Masked Card                       : {masked_card}")
```

CUSTOMER INFORMATION PROCESSOR (RETAIL ANALYTICS)

```python
# --- Creating variable and using string slicing for customer record processing
customer_record = "CUS-LAG-2026-000451"

print("Customer Prefix       :", customer_record[0:3])
print("City Code             :", customer_record[4:7])
print("Registration Year     :", customer_record[8:12])
print("Customer Number       :", customer_record[13:19])
print("First 8 Characters    :", customer_record[0:8])
print("Last 5 Characters     :", customer_record[-5:])
print("Every Second Character:", customer_record[::2])
print("Every Third Character :", customer_record[::3])
print("Reverse Record        :", customer_record[::-1])

# Step 3: Customer Name Processing
customer_name = "Christopher Johnson"

print("First Name            :", customer_name[0:11])
print("Last Name             :", customer_name[12:19])
print("First 5 Letters       :", customer_name[0:5])
print("Last 7 Letters        :", customer_name[-7:])
print("Reversed Name         :", customer_name[::-1])
```
